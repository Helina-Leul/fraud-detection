```python
import pandas as pd

fraud = pd.read_csv("../data/raw/Fraud_Data.csv")

fraud['signup_time'] = pd.to_datetime(fraud['signup_time'])
fraud['purchase_time'] = pd.to_datetime(fraud['purchase_time'])

fraud.head()
```

```python
fraud['time_since_signup'] = (
    fraud['purchase_time'] - fraud['signup_time']
).dt.total_seconds()
```

```python
fraud['hour_of_day'] = fraud['purchase_time'].dt.hour
```

```python
fraud['day_of_week'] = fraud['purchase_time'].dt.dayofweek
```

```python
fraud['user_transaction_count'] = fraud.groupby('user_id')['user_id'].transform('count')
```

```python
fraud['device_transaction_count'] = fraud.groupby('device_id')['device_id'].transform('count')
```

```python
fraud['ip_address_int'] = fraud['ip_address'].astype(int)
```

```python
fraud[['time_since_signup',
       'hour_of_day',
       'day_of_week',
       'user_transaction_count',
       'device_transaction_count']].head()
```

Feature engineering was applied to extract behavioral signals from raw transaction data. Time-based features such as time_since_signup and hour_of_day capture temporal fraud patterns, while frequency-based features such as user_transaction_count and device_transaction_count help identify abnormal repeated behavior often associated with fraudulent activity.
